# 1장 딥러닝과 파이토치 라이브러리 소개
우리가 말하는 일반화된 알고리즘 분류는 AI의 하위 범주인 딥러닝에 해당.

딥러닝은 가르치기에 적합한 예제들을 사용하여 심층 신경망 이라는 수학적 개체를 훈련시키는 일을 한다.

딥러닝은 대량의 데이터를 사용해 입력과 출력이 동떨어진 복잡한 함수들을 근사하는 방법을 사용.

딥러닝 덕분에 우리는 최근까지 사람만이 독점적으로 행할 수 있던 기능을 프로그램에도 부여할 수 있게 됨.

## 1.1 딥러닝 혁명
머신러닝 - 피처 엔지니어링에 크게 의존.

| 피처 엔지니어링은 분류기 알고리즘이 새로운 데이터를 기반으로 올바른 결과물을 촉진하기 위해 입력 데이터를 변환하는 작업

예를 들어 손글씨 숫자의 0과 1의 이미지를 구분해내려면, 이미지의 테두리선 분포를 추정하기 위한 필터 집합을 구성.

주어진 테두리선의 방향 분포로 올바른 숫자를 예측하게끔 분류기를 훈련.

반면, 딥러닝은 숫자 인식 작업을 성공하기 위해 원본 데이터로부터 자동으로 표현들을 찾아냄. 반복훈련 결과 자동으로 개선

학습 시스템에는 어떤 형태로든 사전 지식을 주입할 필요가 있기에 피처엔지니어링도 사용하고, 데이터가 주어진 예제로 부터 유용한 표현을 추출하는 신경망의 능력으로 딥러닝도 강력하게 사용한다.

신경망 내부의 수학적 개체를 조작해 훈련 데이터로부터 표현을 자동으로 훈련시키는데 집중한다.

### 피처 엔지니어링
실무자가 피처를 잘 만들어낼수록 학습도 좋은 결과를 가져다 줌.

`데이터->수작업생성 피처->머신 러닝->결과`

### 딥러닝
원본데이터가 알고리즘에 바로 들어가고 작업 절차에서 성능 최적화의 유도 과정에 따라 저수준에서 고수준의 피처들이 자동으로 추출.

`데이터->딥러닝 머신->찾아낸 표현들->결과`

@딥러닝을 성공적으로 실행하기 위한 필요한 요구사항
- 가지고 있는 데이터가 무엇이든 넣을 수 있어야 한다
- 딥러닝 머신을 정의
유용한 표현을 얻고 머신이 원하는 출력 결과를 낼 수 있도록 자동화된 훈련 방법을 생성


---

`훈련 과정`에서는 **기대한 값**과 **실제 모델 출력의 차이**를 점수로 제공.

그렇기 위해 참조 데이터 그리고 모델의 출력값을 사용한 **실수 함수로 기준**을 만들어 사용.

통상적으로 점수는 낮을수록 좋고, 훈련은 기준을 통해 **점점 더 낮은 점수를 얻도록 점진적**으로 딥러닝 머신을 변경하여, 훈련 중에 입력되지 않은 데이터에 대해서도 낮은 점수를 달성하게 만듬





## 1.2 딥러닝을 위한 파이토치
### 파이토치
명확한 문법과 현대적인 API, 쉬운 디버깅 - 딥러닝 도입을 위한 최선의 선택

딥러닝 머신 - `입력을 출력으로 매핑하는 다소 복잡한 수학 함수`

파이토치 텐서 - 핵심 데이터 구조를 제공, 텐서는 넘파이 배열과 여러 면에서 유사한 다차원 배열
- 자료주고 위에서 파이토치는 전용 하드웨어를 십분 활용해 수학 연산을 빠르게 수행.
- 신경망 아키텍처 설꼐, 개별 혹은 분산 컴퓨팅 자원에서 훈련하기 매우 편리한 장점

### 커리큘럼 구성(총 3부)
1. 1부) 기본 개념 설명 - 파이토치 제공 기능, 딥러닝 실제 코드
2. 2부) 의학 영상 자료 전체 프로젝트(기본 개념 토대로 CT 사진 종양 식별 예제)
3. 3부) 딥러닝 모델의 실제 적용을 위한 파이토치 제공 배포 기능 소개

파이토치를 이용해 2차원-3차원 데이터셋 이미지를 처리해서 소규모로 분류하고 분할하는 예제가 흥미롭다.



## 1.3 왜 파이토치인가?
딥러닝은 다양한 분야의 복잡한 작업을 수행하게 함.

기계 번역이나 전략 게임 실행 혹은 어수선한 장면에서의 물체 인식 등 작업은 모두 **우리가 만든 모델을 여러 예제 데이터에 노출해 학습**시켜줘야 가능하다.

실제 학습을 위해서는 다양한 문제에 적용할 수 있을 정도로 유연하면서도, **적당한 시간 안에 많은 양의 데이터로 훈련 가능한 효율적인 도구가 필요**하다. 또 이 도구로 훈련된 모델은 **다양한 입력에 대해 정확하게 동작**해야 한다.

### 파이토치 추천하는 이유
1. 단순함. 학습이나 활용, 확장, 디버깅 면에서 모두 수월, 파이토치는 파이썬스러운 면이 있다.
2. 파이토치로 딥러닝 머신을 프로그래밍하는 것은 정말 자연스러움, 숫자나 벡터 혹은 행렬이나 배열을 담을 수 있는 데이터 타입인 `Tensor`을 기본으로 제공.
  - 부가적으로 텐서 연산을 위한 함수도 지원.(넘파이 활용이 큼)

### 파이토치 딥러닝 기능(과학 계산 최적화를 지원하는 고성능 라이브러리)
1. 그래픽 연산 유닛(GPU)로 연산을 가속
2. 딥러닝이 사용하는 일반 수학식에 대해 산술 최적화 지원

파이토치는 딥러닝 영역의 아이디어를 파이썬 코드로 즉시 매끈하게 만들어낼 수 있는 구현 도구.

또한, 파이토치는 C++로 모델을 설계하고 훈련시키는 과정에서도 사용 가능.(파이토치는 C++과 연동이좋음)

다른 언어를 지원하는 바인딩과 모바일 장비에 배포하기 위한 인터페이스도 늘린다.


### 파이토치와 텐서플로우의 차이
| **항목**     | **PyTorch**                                                                        | **TensorFlow**                                                                      |
| ---------- | ---------------------------------------------------------------------------------- | ----------------------------------------------------------------------------------- |
| **장점**     | - **직관적이고 Pythonic** (즉시 실행: Eager Execution)  - 디버깅과 코드 테스트가 쉬움  - 연구자/학계에서 많이 사용 | - **생산 배포에 최적화**  - TensorFlow Serving, TFLite 등 **배포 도구 풍부**  - Google의 강력한 생태계 지원 |
| **단점**     | - 배포 측면에서는 상대적으로 미흡  - 모바일/임베디드 환경에 약함                                             | - 복잡한 그래프 정의 방식 (1.x 버전 기준)  - 디버깅이 불편할 수 있음                                        |
| **주요 사용처** | 연구, 프로토타이핑, 실험                                                                     | 산업 현장, 대규모 배포, 모바일                                                                  |

## 1.4 파이토치 딥러닝 프로젝트 둘러보기
파이썬 API - 놀라운 활용성과 파이썬 생태계와의 폭넓은 통합 이면

파이토치는 기본적으로 `다차원 배열` 혹은 `텐서` 라 부르는 자료구조를 사용하는 라이브러리.

텐서를 사용하는 다양한 연산 라이브러리를 `torch` 모듈로 제공, 텐서 자료구조나 이를 사용한 연산은 `CPU` 나 `GPU` 에서 수행.(한 두개 함수로 호출)

### 자동미분
특정 텐서에서 수행한 모든 연산을 기억해줬다가 주어진 입력값을 기준으로 미분값을 자동 계산(자동미분), 이처럼 파이토치의 자동미분이 텐서를 사용하기 때문에 가능.(파이토치의 계산 최적화 기능)
- 물리 연산, 렌더링, 최적화, 시뮬레이션, 모델링 등 사용
- 신경망을 만들고 훈련하는데 필요한 모든 빌딩 블럭을 제공

### 파이토치 프로젝트의 기본구조(데이터 로딩, 훈련, 제품화 등)
```
데이터소스 -> 샘플텐서(다중프로세스 데이터 로딩)->배치 텐서
-> 훈련 루프(훈련전 모델) - 여러 대의 서버/GPU에서 분산 훈련
-> 훈련된 모델
-> 출시용 모델(ONNX, JIT 토치스크립트) - 클라우드/운영서버
```

### 신경망 구축을 지원하는 파이토치 핵심 모듈
`torch.nn` - 공통적인 신경망과 아키텍처적인 구성 요소 제공.(완전연결계층,컨볼루션층,활성화함수,손실함수 등 포함) 이들은 훈련전 모델을 만들고 초기화하는데 필요

모델 훈련을 위해 훈련 데이터를 얻을 수 있는 곳, 모델을 훈련 데이터에 맞춰주는 옵티마이저, 모델과 데이터로 모델 훈련을 위해 필요한 계산을 수행할 하드웨어 등 필요

`torch.utils.data -> dataset 클래스` - 입수한 데이터를 표준화된 파이토치 텐서로 전환(저장소를 비롯한 데이터 출처로 부터 물리적으로 데이터를 가져오는 작업이 필요, 데이터에서 가져온 각각의 샘플을 파이토치가 다룰 수 있는 텐서로 변환.)

`DataLoader 클래스` - 데이터 저장소는 대부분 느려 병열로 로딩, 하지만 여러 개의 훈련 샘플을 묶어놓은 텐서 형태인 `배치(batch)`로 데이터를 묶기 위해 병렬처리가 필요.
- 알아서 자식 프로세스가 뜨고 백그라운드로 동작, 훈련 루프가 필요할 경우 바로 전달하도록 준비

`훈련루프`는 매번 데이터 로더에서 얻은 샘플로 모델을 평가.

`torch.nn -> 손실함수` - 출력 결과와 예측 결과와 비교(모델의 출력을 기대값(목표값)과 비교하는데 이때 기준 혹은 손실 함수를 사용한다. 즉 필요한 요소뿐만 아니라 사용가능한 손실 함수도 제공한다.)
- 다음 출력이 목표 출력을 더 잘 모사하게끔 모델 조정, 자동미분 엔진을 사용.

`torch.optim` - 조정을 수행하는 옵티마이저도 필요

`torch.nn.parallel.DistributedDataParallel`과 `torch.distributed` 서브 모듈 - 복잡한 하드웨어 사용(더 큰 모델을 훈련시킬 목적으로 자우너을 제공하는 여러 GPU나 여러 컴퓨터같은 복잡한 하드웨어를 점점 더 많이 사용)

---

훈련루프 과정 이후 모델이 원하는 작업에 맞게 최적화되는 보상을 얻는다.

`토치스크립트` - 모델을 미리 컴파일하는 방식도 제공, 토치스크립트를 사용하면 모델을 직렬화해서 여러 명령어 집합으로 바꾸고 나중에 파이썬에 의존하지 않고서 C++ 프로그램이나 모바일 기기에서 실행 가능

텐서 연산과 연관된 **제한된 명령어 집합을 지원**하는 가상 머신이다.

이 기능은 모델을 `토치스크립트` / `ONNX(표준화된 포맷)` 으로 내보내어 사용 가능하여 **실서비스 배포**를 위한 기본 기능
