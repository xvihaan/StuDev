```python
from torchvision import transforms
preprocess = transforms.Compose([
        transforms.Resize(256),
        transforms.CenterCrop(224),
        transforms.ToTensor(),
        transforms.Normalize(
            mean=[0.485, 0.456, 0.406],
            std=[0.229, 0.224, 0.225]
        )])
```

```python
from PIL import Image
img = Image.open("/content/image/KakaoTalk_Photo_2025-01-23-16-20-24.png").convert('RGB') # Convert the image to RGB
```

```python
img
```

```python
img_t = preprocess(img)
```

```python
import torch
batch_t = torch.unsqueeze(img_t, 0)
```

```python
resnet.eval()
```

```python
out = resnet(batch_t)
```

```python
out
```

이전 과정에서 얻은 out 텐서에게 가장 높은 점수에 대응하는 색인을 찾아야 한다. 파이토치의 max 함수를 사용해 텐서에서 최댓값과 이 최댓값이 들어 있는 부분의 인덱스를 출력

```python
with open('/content/imagenet_classes.txt') as f:
    labels = [line.strip() for line in f.readlines()]
```

```python
_, index = torch.max(out, 1)
```

레이블에 접근하기 위해 직전에 출력한 인덱스 사용.

인덱스가 단순한 파이썬 숫자가 아닌, 1개의 요소를 가지는 1차원 텐서(tensor[207]) 이다. 그래서 labels 리스트의 인덱스로 사용하기를 원하는 실제 정수 값을 얻기 위해 Index[0]을 사용할 수가 있다. 그리고 출력을 [0, 1] 사이의 값으로 정규화 하고 전체 합으로 나누기 위해(torch.nn.functional.softmax) 사용

```python
percentage = torch.nn.functional.softmax(out, dim=1)[0] * 100
```

```python
labels[index[0]], percentage[index[0]].item()
```

모델은 여러 점술

모델은 여러 점수를 출력하므로, 두-세번째의 결과가 무엇인지 찾아낼 수 있다. sort 함수 사용, 배열 값 기준으로 오름차순/내림차순으로 정렬한 후 값과 인덱스를 보여준다.

```python
_, indices = torch.sort(out, descending=True)
[(labels[idx], percentage[idx].item()) for idx in indices[0][:5]]
```
